# Exploration
Quick look at the DuckDB tables produced by `make all`. All tables live in `data/nfl.duckdb`.

```python
import duckdb
import polars as pl

con = duckdb.connect('../data/nfl.duckdb', read_only=True)
con.sql('SHOW TABLES').pl()
```

## Archetypes

```python
con.sql('SELECT * FROM archetype_centroids_wr ORDER BY n DESC').pl()
```

```python
con.sql("SELECT player_name, season, team, archetype, round(membership_prob, 2) AS p\nFROM archetype_qb WHERE season = 2025 ORDER BY dropbacks DESC LIMIT 15").pl()
```
